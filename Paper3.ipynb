```python
import tensorflow as tf
import os
import matplotlib.pyplot as plt
import random
import glob
from tensorflow_model_optimization.python.core.keras.compat import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np

# Directory where the dataset is stored
output_dir = '/content/Merged_Split'  # Where to save the split dataset

# Set image dimensions
BATCH_SIZE = 16
IMG_DIM = 51
NUM_CLASSES = 6

# Color adjustment augmentation pipeline
color_augmentation = ImageDataGenerator(
    rescale=1.0 / 255,  # Normalize pixel values to [0, 1]
    brightness_range=[0.3, 1.5],  # Random brightness adjustment
    zoom_range=0.2,               # Random zoom
    channel_shift_range=20.0      # Random channel shift
)

# Shifting, rotation, flips augmentation pipeline
shift_rotate_flip_augmentation = ImageDataGenerator(
    rescale=1.0 / 255,  # Normalize pixel values to [0, 1]
    rotation_range=30,   # Random rotation
    width_shift_range=0.2,  # Random horizontal shift
    height_shift_range=0.2,  # Random vertical shift
    shear_range=0.2,     # Shear transformation
    horizontal_flip=True,  # Random horizontal flip
    vertical_flip=True     # Random vertical flip
)

# Contrast, Zoom, Flip augmentation pipeline (Removed ZCA Whitening)
contrast_zoom_flip_augmentation = ImageDataGenerator(
    rescale=1.0 / 255,  # Normalize pixel values to [0, 1]
    featurewise_center=True,  # Normalize image to have zero mean
    featurewise_std_normalization=True,  # Normalize by the standard deviation
    channel_shift_range=40.0,  # Randomly shift colors of the channels
    cval=0.0  # Value to fill in for pixels outside the boundaries during transformations
)

# Load the dataset and get class indices
train_generator = color_augmentation.flow_from_directory(
    f"{output_dir}/train",
    target_size=(IMG_DIM, IMG_DIM),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

class_indices = train_generator.class_indices  # Get class-to-index mapping

# Function to visualize original vs augmented images
def plot_augmented_images(datagen, class_name, class_path, augmentation_type):
    img_paths = glob.glob(f"{class_path}/*.*")  # Match all image types
    if not img_paths:
        print(f"Warning: No images found in {class_path}")
        return

    random_image_path = random.choice(img_paths)
    img = keras.preprocessing.image.load_img(random_image_path, target_size=(IMG_DIM, IMG_DIM))
    img_array = keras.preprocessing.image.img_to_array(img)
    img_array = img_array.reshape((1,) + img_array.shape)  # Reshape for batch processing

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    # Display the original image
    axes[0].imshow(img)
    axes[0].set_title(f"Original: {class_name}")
    axes[0].axis("off")

    # Generate and display augmented images
    i = 1
    for batch in datagen.flow(img_array, batch_size=1):
        axes[i].imshow(batch[0])
        axes[i].set_title(f"{augmentation_type} Augmented")
        axes[i].axis("off")
        i += 1
        if i == 3:
            break
    plt.show()

# Loop over each class and display original vs augmented images
for class_name in sorted(class_indices.keys()):
    print(f"\nShowing augmented images for class: {class_name}")

    # Display images for color adjustment pipeline
    print(f"\n{class_name} - Color Adjustments Pipeline:")
    plot_augmented_images(color_augmentation, class_name, f"{output_dir}/train/{class_name}", "Color Adjustments")

    # Display images for shifting, rotation, flips pipeline
    print(f"\n{class_name} - Shifting, Rotation, Flips Pipeline:")
    plot_augmented_images(shift_rotate_flip_augmentation, class_name, f"{output_dir}/train/{class_name}", "Shifting/Rotation/Flips")

    # Display images for contrast, zoom, flip pipeline
    print(f"\n{class_name} - Contrast, Zoom, Flip Pipeline:")
    plot_augmented_images(contrast_zoom_flip_augmentation, class_name, f"{output_dir}/train/{class_name}", "Contrast/Zoom/Flip")
```

```python
import os
from tensorflow_model_optimization.python.core.keras.compat import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tqdm.notebook import tqdm  # For progress bar
import glob

# Directory where the augmented images will be saved
augmented_dir = '/content/augmented_train_images'  # Update to your preferred directory
os.makedirs(augmented_dir, exist_ok=True)

# Augmentation pipelines (already defined in previous code)
color_augmentation = ImageDataGenerator(
    rescale=1.0 / 255,
    brightness_range=[0.3, 1.5],
    zoom_range=0.2,
    channel_shift_range=20.0
)

shift_rotate_flip_augmentation = ImageDataGenerator(
    rescale=1.0 / 255,
    rotation_range=30,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    horizontal_flip=True,
    vertical_flip=True
)

contrast_zoom_flip_augmentation = ImageDataGenerator(
    rescale=1.0 / 255,
    featurewise_center=True,
    featurewise_std_normalization=True,
    zca_whitening=True,
    zca_epsilon=1e-06,
    channel_shift_range=40.0,
    cval=0.0,
    interpolation_order=6
)

# Function to save augmented images for a given pipeline
def save_augmented_images(datagen, class_name, class_path):
    # Create a directory to save the augmented images for each class
    class_augmented_dir = os.path.join(augmented_dir, class_name)
    os.makedirs(class_augmented_dir, exist_ok=True)

    # Get all images in the class folder
    img_paths = glob.glob(f"{class_path}/*.jpg")

    # Loop through images and apply augmentation
    for i, img_path in enumerate(tqdm(img_paths, desc=f"Saving augmented images for {class_name}", unit="image")):
        img = keras.preprocessing.image.load_img(img_path, target_size=(IMG_DIM, IMG_DIM))
        img_array = keras.preprocessing.image.img_to_array(img)
        img_array = img_array.reshape((1,) + img_array.shape)  # Reshape for batch processing

        # Apply the augmentation and save images
        for batch in datagen.flow(img_array, batch_size=1, save_to_dir=class_augmented_dir, save_prefix=f"{class_name}_aug", save_format="jpeg"):
            break  # Save one augmented image per original image

# Apply and save augmented images for each class using the three augmentation pipelines
for class_name in sorted(class_indices.keys()):
    print(f"\nProcessing class: {class_name}")

    # Save images for color adjustment pipeline
    save_augmented_images(color_augmentation, class_name, f"{output_dir}/train/{class_name}")

    # Save images for shifting, rotation, flips pipeline
    save_augmented_images(shift_rotate_flip_augmentation, class_name, f"{output_dir}/train/{class_name}")

    # Save images for contrast, zoom, flip pipeline
    save_augmented_images(contrast_zoom_flip_augmentation, class_name, f"{output_dir}/train/{class_name}")

print("\nAugmented images saved successfully!")
```

```python
import random
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

# Set random seed for reproducibility
SEED = 42
tf.random.set_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)

# Image dimensions and number of classes
IMG_DIM = 51
NUM_CLASSES = 6

# Define RepVGG Block
class RepVGGBlock(tf.keras.layers.Layer):
    def __init__(self, filters, kernel_size=3, stride=1, use_identity=True):
        super(RepVGGBlock, self).__init__()
        self.conv3x3 = layers.Conv2D(filters, kernel_size, strides=stride, padding='same', use_bias=False)
        self.bn3x3 = layers.BatchNormalization()
        self.conv1x1 = layers.Conv2D(filters, 1, strides=stride, padding='same', use_bias=False)
        self.bn1x1 = layers.BatchNormalization()

        if use_identity:
            if stride > 1:
                self.identity = layers.Conv2D(filters, 1, strides=stride, padding="same", use_bias=False)
            else:
                self.identity = layers.BatchNormalization()
        else:
            self.identity = None

        self.relu = layers.ReLU()

    def call(self, inputs, training=False):
        x1 = self.relu(self.bn3x3(self.conv3x3(inputs), training=training))
        x2 = self.relu(self.bn1x1(self.conv1x1(inputs), training=training))

        if self.identity is not None:
            x3 = self.identity(inputs)
            return self.relu(x1 + x2 + x3)
        else:
            return self.relu(x1 + x2)

# Define Multi-Scale Convolutional Attention (MSCA) Module
class MSCA_Module(tf.keras.layers.Layer):
    def __init__(self, filters):
        super(MSCA_Module, self).__init__()
        self.conv1 = layers.Conv2D(filters // 2, kernel_size=1, activation='relu', padding='same')
        self.conv3 = layers.Conv2D(filters // 2, kernel_size=3, activation='relu', padding='same')
        self.conv5 = layers.Conv2D(filters, kernel_size=5, activation='relu', padding='same')
        self.bn = layers.BatchNormalization()

    def call(self, inputs, training=False):
        x1 = self.conv1(inputs)
        x2 = self.conv3(inputs)
        x3 = self.conv5(inputs)
        x = tf.concat([x1, x2, x3], axis=-1)
        x = self.bn(x, training=training)
        return x

# Build RepVGG-MEM Model
def build_RepVGG_MEM(input_shape=(IMG_DIM, IMG_DIM, 3), num_classes=NUM_CLASSES):
    inputs = layers.Input(shape=input_shape)

    x = RepVGGBlock(64, stride=2)(inputs)    # Output: (26, 26, 64)
    x = RepVGGBlock(128, stride=2)(x)        # Output: (13, 13, 128)
    x = RepVGGBlock(256, stride=2)(x)        # Output: (7, 7, 256)

    x = MSCA_Module(256)(x)                  # Apply MSCA Module
    x = layers.GlobalAveragePooling2D()(x)   # Output: (256,)

    x = layers.Dense(64, activation='relu')(x)
    x = layers.Dropout(0.5)(x)
    x = layers.Dense(32, activation='relu')(x)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = models.Model(inputs, outputs)
    return model

# Instantiate and compile the model
tf_model = build_RepVGG_MEM()
tf_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Print model summary
tf_model.summary()
```

```python
from tensorflow.keras.callbacks import Callback, ModelCheckpoint, EarlyStopping
from sklearn.metrics import f1_score as sklearn_f1_score
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

# Custom Progress Bar Callback with integrated F1 Score
class CustomProgressCallback(Callback):
    def __init__(self, validation_data, total_epochs):
        super().__init__()
        self.validation_data = validation_data  # (val_images, val_labels)
        self.epochs = total_epochs
        self.f1_scores = []
        self.pbar = None

    def on_train_begin(self, logs=None):
        self.pbar = tqdm(total=self.epochs, desc="Training")

    def on_epoch_begin(self, epoch, logs=None):
        self.epoch_pbar = tqdm(total=steps_per_epoch_train, desc=f"Epoch {epoch+1}/{self.epochs}",
                              leave=False)

    def on_batch_end(self, batch, logs=None):
        self.epoch_pbar.update(1)
        # Update metrics for current batch
        metrics_str = " - ".join(f"{k}: {v:.4f}" for k, v in logs.items())
        self.epoch_pbar.set_postfix_str(metrics_str)

    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        val_images, val_labels = self.validation_data
        val_preds = self.model.predict(val_images, verbose=0)
        val_pred_classes = np.argmax(val_preds, axis=1)
        val_true_classes = np.argmax(val_labels, axis=1)

        f1 = sklearn_f1_score(val_true_classes, val_pred_classes, average='weighted')
        self.f1_scores.append(f1)
        logs['val_f1_score'] = f1

        # Close epoch progress bar
        self.epoch_pbar.close()

        # Update main progress bar with all metrics
        metrics_str = " - ".join(f"{k}: {v:.4f}" for k, v in logs.items())
        self.pbar.set_postfix_str(metrics_str)
        self.pbar.update(1)

    def on_train_end(self, logs=None):
        self.pbar.close()

# Calculate steps per epoch for training and validation
steps_per_epoch_train = np.ceil(train_dataset_tf.samples / BATCH_SIZE).astype(int)
steps_per_epoch_val = np.ceil(val_dataset_tf.samples / BATCH_SIZE).astype(int)

# Load validation data into memory
val_X, val_y = [], []
val_dataset_tf.reset()  # Reset the generator before using it
for _ in range(steps_per_epoch_val):
    try:
        batch_X, batch_y = next(val_dataset_tf)
        val_X.append(batch_X)
        val_y.append(batch_y)
    except StopIteration:
        break

val_X = np.vstack(val_X)
val_y = np.vstack(val_y)

# Set up callbacks
checkpoint = ModelCheckpoint(
    '/content/drive/MyDrive/Colab Notebooks/best_model.h5',
    monitor='val_accuracy',
    save_best_only=True,
    mode='max',
    verbose=1
)

early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=20,
    restore_best_weights=True,
    verbose=1
)

# Define number of epochs
epochs = 50

# Create custom progress callback
progress_callback = CustomProgressCallback(validation_data=(val_X, val_y), total_epochs=epochs)

# Train the model
history = tf_model.fit(
    train_dataset_tf,
    steps_per_epoch=steps_per_epoch_train,
    epochs=epochs,
    validation_data=val_dataset_tf,
    validation_steps=steps_per_epoch_val,
    callbacks=[
        progress_callback,  # Custom progress bar with F1 score
        checkpoint,         # Model checkpoint
        early_stopping,     # Early stopping
    ],
    verbose=0  # Turn off default verbose output
)

# Save training metrics to CSV
history_dict = history.history
num_epochs = len(history_dict['loss'])

# Add val_f1_score from callback if not already in history
if 'val_f1_score' not in history_dict:
    history_dict['val_f1_score'] = progress_callback.f1_scores

# Construct DataFrame
history_df = pd.DataFrame({
    'epoch': range(1, num_epochs + 1),
    'accuracy': history_dict['accuracy'],
    'loss': history_dict['loss'],
    'val_accuracy': history_dict['val_accuracy'],
    'val_loss': history_dict['val_loss'],
    'val_f1_score': progress_callback.f1_scores[:num_epochs]
})

# Save CSV
csv_path = '/content/drive/MyDrive/paper3/training_metrics.csv'
history_df.to_csv(csv_path, index=False)
print(f"\nTraining metrics saved to {csv_path}")

# Final evaluation and model save
test_loss, test_accuracy = tf_model.evaluate(test_dataset_tf, verbose=1)
print(f'Test Accuracy: {test_accuracy * 100:.2f}%')

# Save the model
tf_model.save('/content/saved_model.h5', save_format='h5')  # Explicitly set format to h5
print("\nModel training complete and saved to /content/saved_model.h5")

# Visualize training history
plt.figure(figsize=(15, 5))

# Plot training & validation accuracy
plt.subplot(1, 3, 1)
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='lower right')

# Plot training & validation loss
plt.subplot(1, 3, 2)
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper right')

# Plot F1 score
plt.subplot(1, 3, 3)
plt.plot(progress_callback.f1_scores)
plt.title('Validation F1 Score')
plt.ylabel('F1 Score')
plt.xlabel('Epoch')
plt.legend(['Validation'], loc='lower right')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import os
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.metrics import confusion_matrix, roc_curve, auc, classification_report, accuracy_score
from itertools import cycle
from PIL import Image

# Set the constants needed for testing
IMG_DIM = 51
NUM_CLASSES = 6

# Define custom layers needed for model loading with proper **kwargs handling
class RepVGGBlock(tf.keras.layers.Layer):
    def __init__(self, filters, kernel_size=3, stride=1, use_identity=True, **kwargs):
        super(RepVGGBlock, self).__init__(**kwargs)
        self.filters = filters
        self.kernel_size = kernel_size
        self.stride = stride
        self.use_identity = use_identity
        self.conv3x3 = layers.Conv2D(filters, kernel_size, strides=stride, padding='same', use_bias=False)
        self.bn3x3 = layers.BatchNormalization()
        self.conv1x1 = layers.Conv2D(filters, 1, strides=stride, padding='same', use_bias=False)
        self.bn1x1 = layers.BatchNormalization()

        if use_identity:
            if stride > 1:
                self.identity = layers.Conv2D(filters, 1, strides=stride, padding="same", use_bias=False)
            else:
                self.identity = layers.BatchNormalization()
        else:
            self.identity = None

        self.relu = layers.ReLU()

    def call(self, inputs, training=False):
        x1 = self.relu(self.bn3x3(self.conv3x3(inputs), training=training))
        x2 = self.relu(self.bn1x1(self.conv1x1(inputs), training=training))

        if self.identity is not None:
            x3 = self.identity(inputs)
            return self.relu(x1 + x2 + x3)
        else:
            return self.relu(x1 + x2)

    def get_config(self):
        config = super(RepVGGBlock, self).get_config()
        config.update({
            'filters': self.filters,
            'kernel_size': self.kernel_size,
            'stride': self.stride,
            'use_identity': self.use_identity
        })
        return config

class MSCA_Module(tf.keras.layers.Layer):
    def __init__(self, filters, **kwargs):
        super(MSCA_Module, self).__init__(**kwargs)
        self.filters = filters
        self.conv1 = layers.Conv2D(filters // 2, kernel_size=1, activation='relu', padding='same')
        self.conv3 = layers.Conv2D(filters // 2, kernel_size=3, activation='relu', padding='same')
        self.conv5 = layers.Conv2D(filters, kernel_size=5, activation='relu', padding='same')
        self.bn = layers.BatchNormalization()

    def call(self, inputs, training=False):
        x1 = self.conv1(inputs)
        x2 = self.conv3(inputs)
        x3 = self.conv5(inputs)
        x = tf.concat([x1, x2, x3], axis=-1)
        x = self.bn(x, training=training)
        return x

    def get_config(self):
        config = super(MSCA_Module, self).get_config()
        config.update({
            'filters': self.filters
        })
        return config

# Option 1: Try to load with custom objects
try:
    print("Attempting to load model with custom object scope...")
    model_path = "/content/saved_model.h5"
    custom_objects = {
        'RepVGGBlock': RepVGGBlock,
        'MSCA_Module': MSCA_Module
    }

    with tf.keras.utils.custom_object_scope(custom_objects):
        tf_model = tf.keras.models.load_model(model_path)
    print("✅ Model loaded successfully with custom object scope.")
except Exception as e:
    print(f"Error loading model with custom object scope: {e}")

    # Option 2: Rebuild the model and load weights
    print("\nAttempting to rebuild model and load weights...")

    # Rebuild the model architecture
    def build_RepVGG_MEM(input_shape=(IMG_DIM, IMG_DIM, 3), num_classes=NUM_CLASSES):
        inputs = layers.Input(shape=input_shape)

        x = RepVGGBlock(64, stride=2)(inputs)
        x = RepVGGBlock(128, stride=2)(x)
        x = RepVGGBlock(256, stride=2)(x)

        x = MSCA_Module(256)(x)
        x = layers.GlobalAveragePooling2D()(x)

        x = layers.Dense(64, activation='relu')(x)
        x = layers.Dropout(0.5)(x)
        x = layers.Dense(32, activation='relu')(x)
        x = layers.Dropout(0.5)(x)
        outputs = layers.Dense(num_classes, activation='softmax')(x)

        model = models.Model(inputs, outputs)
        return model

    # Instantiate and compile the model
    tf_model = build_RepVGG_MEM()
    tf_model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    # Try to load weights
    try:
        tf_model.load_weights(model_path)
        print("✅ Model rebuilt and weights loaded successfully.")
    except Exception as e:
        print(f"Error loading weights: {e}")
        print("\nFalling back to training weights...")
        # If weights loading fails, try weights directory
        try:
            tf_model.load_weights("/content/drive/MyDrive/Colab Notebooks/best_model.h5")
            print("✅ Model weights loaded from training checkpoint.")
        except Exception as e2:
            print(f"Error loading weights from checkpoint: {e2}")
            print("⚠️ Proceeding with uninitialized model. Results may be incorrect.")

# Load test dataset
test_data_dir = "/content/Merged_Split/test"
test_datagen = tf.keras.preprocessing.image.ImageDataGenerator(rescale=1.0 / 255)

test_dataset_tf = test_datagen.flow_from_directory(
    test_data_dir,
    target_size=(IMG_DIM, IMG_DIM),
    batch_size=32,
    class_mode='categorical',
    shuffle=False
)

# Get true labels and class names
true_classes = test_dataset_tf.classes
class_names = list(test_dataset_tf.class_indices.keys())
num_classes = len(class_names)

# Evaluate test accuracy using model.evaluate()
print("\n📊 Evaluating model on test dataset...")
test_loss, test_accuracy = tf_model.evaluate(test_dataset_tf, verbose=1)
print(f"\n📈 Test Accuracy: {test_accuracy * 100:.2f}%")
print(f"📉 Test Loss: {test_loss:.4f}")

# 1. Predictions on Test Data
y_pred_probs = tf_model.predict(test_dataset_tf)
y_pred_classes = np.argmax(y_pred_probs, axis=1)

# Calculate accuracy using scikit-learn for verification
sklearn_accuracy = accuracy_score(true_classes, y_pred_classes)
print(f"📊 Test Accuracy (from predictions): {sklearn_accuracy * 100:.2f}%")

# 2. Classification Report
report = classification_report(true_classes, y_pred_classes, target_names=class_names)
print("\n📄 Classification Report:\n", report)

# Save classification report to a file
report_path = "/content/classification_report.txt"
with open(report_path, "w") as f:
    f.write(f"Test Accuracy: {test_accuracy * 100:.2f}%\n")
    f.write(f"Test Loss: {test_loss:.4f}\n\n")
    f.write(report)
print(f"📂 Classification report saved at: {report_path}")

# 3. Confusion Matrix
cm = confusion_matrix(true_classes, y_pred_classes)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title(f'Confusion Matrix (Accuracy: {test_accuracy:.4f})')
conf_matrix_path = "/content/confusion_matrix.png"
plt.savefig(conf_matrix_path)
plt.show()
print(f"📊 Confusion matrix saved at: {conf_matrix_path}")

# 4. Class-wise Accuracy
class_accuracy = [(cm[i, i] / sum(cm[i])) if sum(cm[i]) > 0 else 0 for i in range(num_classes)]
plt.figure(figsize=(12, 6))
plt.bar(range(num_classes), class_accuracy, color='blue')
plt.title('Class-wise Accuracy')
plt.xlabel('Class')
plt.ylabel('Accuracy')
plt.ylim(0, 1)
plt.xticks(range(num_classes), class_names, rotation=45)

for i, v in enumerate(class_accuracy):
    plt.text(i, v + 0.02, f'{v:.2f}', ha='center')

class_accuracy_path = "/content/class_accuracy.png"
plt.savefig(class_accuracy_path)
plt.show()
print(f"📊 Class-wise accuracy plot saved at: {class_accuracy_path}")

# 5. F1 Score Calculation
report_dict = classification_report(true_classes, y_pred_classes, target_names=class_names, output_dict=True)
f1_scores = [report_dict[class_name]["f1-score"] for class_name in class_names]

plt.figure(figsize=(12, 6))
plt.bar(range(num_classes), f1_scores, color='orange')
plt.title('Class-wise F1 Score')
plt.xlabel('Class')
plt.ylabel('F1 Score')
plt.ylim(0, 1)
plt.xticks(range(num_classes), class_names, rotation=45)
for i, v in enumerate(f1_scores):
    plt.text(i, v + 0.02, f'{v:.2f}', ha='center')
f1_score_path = "/content/f1_score.png"
plt.savefig(f1_score_path)
plt.show()
print(f"📊 F1 score plot saved at: {f1_score_path}")

# 6. ROC Curve
y_true_one_hot = tf.keras.utils.to_categorical(true_classes, num_classes)
fpr, tpr, roc_auc = {}, {}, {}

for i in range(num_classes):
    fpr[i], tpr[i], _ = roc_curve(y_true_one_hot[:, i], y_pred_probs[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

plt.figure(figsize=(12, 8))
colors = cycle(['blue', 'red', 'green', 'orange', 'purple', 'brown'])

for i, color in zip(range(num_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2, label=f'Class {class_names[i]} (AUC = {roc_auc[i]:.2f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Multi-class ROC Curve')
plt.legend(loc='lower right')

roc_curve_path = "/content/roc_curve.png"
plt.savefig(roc_curve_path)
plt.show()
print(f"📈 ROC curve saved at: {roc_curve_path}")

# 7. Class Distribution
plt.figure(figsize=(12, 6))
sns.countplot(x=true_classes, palette='pastel')
plt.title('Test Data Distribution')
plt.xlabel('Class')
plt.ylabel('Count')
plt.xticks(range(num_classes), class_names, rotation=45)
plt.tight_layout()
class_dist_path = "/content/class_distribution.png"
plt.savefig(class_dist_path)
plt.show()
print(f"📊 Class distribution plot saved at: {class_dist_path}")

# 8. Create Dashboard Summary
image_paths = [
    conf_matrix_path,
    class_accuracy_path,
    f1_score_path,
    roc_curve_path,
    class_dist_path
]

existing_images = [path for path in image_paths if os.path.exists(path)]

if len(existing_images) >= 2:
    images = [Image.open(img) for img in existing_images]
    widths, heights = zip(*(img.size for img in images))

    # Create a more organized dashboard
    max_width = max(widths)
    total_height = sum(heights)

    # Create a dashboard with images in a single column
    dashboard = Image.new('RGB', (max_width, total_height), (255, 255, 255))
    y_offset = 0

    for img in images:
        dashboard.paste(img, (0, y_offset))
        y_offset += img.height

    dashboard_path = "/content/model_performance_dashboard.png"
    dashboard.save(dashboard_path)
    print(f"📊 Model performance dashboard saved at: {dashboard_path}")

print("\n✅ Testing complete!")
print(f"📊 Final Test Accuracy: {test_accuracy * 100:.2f}%")
```

```python
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping
from tensorflow.keras import backend as K
from tensorflow.keras.callbacks import Callback
from sklearn.metrics import f1_score
import numpy as np

# Custom callback to compute validation F1-score
class F1ScoreCallback(Callback):
    def __init__(self, validation_data):
        super().__init__()
        self.validation_data = validation_data

    def on_epoch_end(self, epoch, logs=None):
        val_images, val_labels = self.validation_data
        val_predict = np.argmax(self.model.predict(val_images), axis=1)
        val_true = np.argmax(val_labels, axis=1)
        val_f1 = f1_score(val_true, val_predict, average='weighted')
        logs['val_f1_score'] = val_f1
        print(f' - val_f1_score: {val_f1:.4f}')

# Define callbacks
checkpoint = ModelCheckpoint('/content/drive/MyDrive/Colab Notebooks/best_model.h5', monitor='val_accuracy', save_best_only=True, mode='max', verbose=1)
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True, verbose=1)
f1_callback = F1ScoreCallback(next(iter(val_dataset_tf)))

# Train the model
EPOCHS = 50  # Adjust as needed
history = tf_model.fit(
    train_dataset_tf,
    validation_data=val_dataset_tf,
    epochs=EPOCHS,
    callbacks=[checkpoint, early_stopping, f1_callback],
    verbose=1
)

# Evaluate on the test set
test_loss, test_accuracy = tf_model.evaluate(test_dataset_tf, verbose=1)
print(f'Test Accuracy: {test_accuracy * 100:.2f}%')
tf_model.save('/content/saved_model.h5')

print("\nModel training complete and saved to /content/saved_model.h5")

```
